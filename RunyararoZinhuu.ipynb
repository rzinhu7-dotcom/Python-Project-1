This is an example of nested conditions because some if statements sit inside other if statements. The page count check is the outer condition, and the price and discount decisions only run inside it. Inside that, the discount check "card_valid == yes" contains another check, "pages >= 20", so the discount applies only when both are true. The condition that must be checked first is the page count (pages >= 1), because a job with zero or negative pages is invalid and should not be priced at all.

```python
# Exercise 2, Question 5(a)-(c)

# Test 1: no ticket, volunteer pass
has_valid_ticket = False
has_volunteer_pass = True
is_accessibility_guest = False
print("Test 1 | Ticket:", has_valid_ticket, "| Volunteer:", has_volunteer_pass, "| Accessibility:", is_accessibility_guest)
if not has_valid_ticket:
    print("Entry denied")
elif has_volunteer_pass or is_accessibility_guest:
    print("Priority queue")
else:
    print("General entry")
print("-" * 40)

# Test 2: ticket, neither priority condition
has_valid_ticket = True
has_volunteer_pass = False
is_accessibility_guest = False
print("Test 2 | Ticket:", has_valid_ticket, "| Volunteer:", has_volunteer_pass, "| Accessibility:", is_accessibility_guest)
if not has_valid_ticket:
    print("Entry denied")
elif has_volunteer_pass or is_accessibility_guest:
    print("Priority queue")
else:
    print("General entry")
print("-" * 40)

# Test 3: ticket, volunteer pass
has_valid_ticket = True
has_volunteer_pass = True
is_accessibility_guest = False
print("Test 3 | Ticket:", has_valid_ticket, "| Volunteer:", has_volunteer_pass, "| Accessibility:", is_accessibility_guest)
if not has_valid_ticket:
    print("Entry denied")
elif has_volunteer_pass or is_accessibility_guest:
    print("Priority queue")
else:
    print("General entry")
print("-" * 40)

# Test 4: ticket, accessibility guest
has_valid_ticket = True
has_volunteer_pass = False
is_accessibility_guest = True
print("Test 4 | Ticket:", has_valid_ticket, "| Volunteer:", has_volunteer_pass, "| Accessibility:", is_accessibility_guest)
if not has_valid_ticket:
    print("Entry denied")
elif has_volunteer_pass or is_accessibility_guest:
    print("Priority queue")
else:
    print("General entry")
print("-" * 40)

# Test 5: no ticket, nothing else
has_valid_ticket = False
has_volunteer_pass = False
is_accessibility_guest = False
print("Test 5 | Ticket:", has_valid_ticket, "| Volunteer:", has_volunteer_pass, "| Accessibility:", is_accessibility_guest)
if not has_valid_ticket:
    print("Entry denied")
elif has_volunteer_pass or is_accessibility_guest:
    print("Priority queue")
else:
    print("General entry")
```

In the priority rule, "or" means a student only needs one of the two things to get into the priority queue. If they have a ticket and either a volunteer pass or accessibility-support registration (or both), they get priority.
```python
# Exercise 3, Test 1: within budget, registered, valid card
# Exercise 3, Part 1 - Input and validation
budget = float(input("Available budget (USD): "))
campus_hours = float(input("Hours on campus: "))
card_valid = input("Valid student card? (yes/no): ").lower()
event_registered = input("Registered for the event? (yes/no): ").lower()

if budget < 0 or campus_hours < 0:
    print("Error: budget and campus hours cannot be negative.")
else:
    # Part 2 - Calculate the plan
    if card_valid == "yes":
        meal_cost = 850 - 850 * 0.10
    else:
        meal_cost = 850

    if campus_hours >= 4:
        shuttle_cost = 250
    else:
        shuttle_cost = 0

    if event_registered == "yes":
        event_cost = 650
    else:
        event_cost = 0

    total_cost = meal_cost + shuttle_cost + event_cost
    print("Meal cost: USD", meal_cost)
    print("Shuttle cost: USD", shuttle_cost)
    print("Event cost: USD", event_cost)
    print("Total planned cost: USD", total_cost)

    # Part 3 - Budget check
    if total_cost <= budget:
        print("Your plan is within budget.")
    else:
        print("Your plan is over budget by USD", total_cost - budget)

    # Part 4 - Decision rule
    if total_cost <= budget and event_registered == "yes":
        print("Recommendation: Attend the event.")
    elif total_cost <= budget:
        print("Recommendation: Check event registration.")
    else:
        print("Recommendation: Review optional costs.")
```
